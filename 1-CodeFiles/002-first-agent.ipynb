### Our First Autogen Agent

```python
from autogen_agentchat.agents import AssistantAgent
from dotenv import load_dotenv
import os , json 
from autogen_ext.models.openai import OpenAIChatCompletionClient

load_dotenv()
api_key= os.getenv("OPEN_ROUTER_API_KEY")
open_api_key= os.getenv("OPEN_AI_KEY")
```

```python
model_client=OpenAIChatCompletionClient(model='gpt-4o',api_key=open_api_key)
assistant=AssistantAgent(name="openaiassistant",model_client=model_client)
try :
    result = await assistant.run(task="Hi there")
except Exception as e:
    print(f'Error >>>>>>>>>>>>>>>>>>>>>  {e.message}')
```

```python
from langchain_huggingface import ChatHuggingFace ,HuggingFaceEndpoint

llm_endpoint = HuggingFaceEndpoint(repo_id="meta-llama/Meta-Llama-3-8B-Instruct"
                                   ,temperature=0.7
                                   ,max_new_tokens=512
                                   ,task="text-generation")

llm_endpoint
```

```python
llm = ChatHuggingFace(llm=llm_endpoint)
llm
query="what is machine learning"
response = llm.invoke(query)

print(response.content[:200])
```
