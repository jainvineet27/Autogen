#### Async Functionality

```python
import asyncio
from asyncio import *
import time 
```

```python
def prepare_coffee():
    print('wakup')
    print('prepare coffied')
    time.sleep(2)

def prepare_baggle():
    print('preapring baggle')
    time.sleep(3)
    print('baggle completed')

start =time.time()
prepare_baggle()
prepare_coffee()  
end = time.time()  
print(f'total duration {end-start}')
```

```python
def main():
    start = time.time()
    prepare_coffee()
    prepare_baggle()
    end= time.time()
    print(f'total time {end-start:.2f}')

main()
```

```python
async def prepare_coffee():
    print('preparing cofee')
    await asyncio.sleep(2)
    print('coffee completed')

async def prepare_baggle():
    print('preparing baggle')
    await asyncio.sleep(3)
    print('baggle completed')

async def main():
    start= time.time()
    await asyncio.gather(prepare_baggle() , prepare_coffee() )   ## bhai agey nhi badna tab tak ye pura funsih na ho jaye 
    end =time.time()
    print(f'total time {end-start}')
```
